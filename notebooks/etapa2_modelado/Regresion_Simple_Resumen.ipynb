# Regresión Lineal Simple - Base Resumen

## Importamos librerías

```python
#Importamos librerias
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from sklearn.linear_model import LinearRegression
```

## Carga de datos

```python
#Lectura del archivo csv
df = pd.read_csv("base_resumen_limpio.csv")
#Rellenamos nulos
df = df.bfill()
df = df.ffill()
#Mostramos las primeras filas
df.head()
```

## Variables numéricas

```python
#Generamos la lista de variables numéricas
numeric_df = df.select_dtypes(['float','int'])  #Devuelve Columnas
Lista_num = numeric_df.columns                  #Devuelve lista de Columnas numéricas
Lista_num
```

## Definimos las variables del modelo

```python
#Variable objetivo (Y)
Variable_y = "Ventas Real"
#Variable independiente (X)
Variable_x = "Leads"
```

## Ajuste del modelo

```python
#Se define model como la función de regresión lineal
model = LinearRegression()
#Ajustamos el modelo con las variables antes declaradas
model.fit(X=df[[Variable_x]], y=df[Variable_y])
#Predecimos los valores de la variable objetivo
y_pred = model.predict(X=df[[Variable_x]])
```

## Coeficientes del modelo

```python
#Obtenemos el intercepto (b0) y la pendiente (b1)
intercepto = model.intercept_
pendiente = model.coef_[0]
print("Intercepto (b0):", intercepto)
print("Pendiente (b1):", pendiente)
#Mostramos la ecuación del modelo
print(f"Ecuación: {Variable_y} = {intercepto:.4f} + ({pendiente:.4f}) * {Variable_x}")
```

## Coeficiente de determinación y correlación

```python
#Corroboramos cual es el coeficiente de Determinación de nuestro modelo
coef_Deter_simple = model.score(X=df[[Variable_x]], y=df[Variable_y])
#Corroboramos cual es el coeficiente de Correlación de nuestro modelo
coef_Correl_simple = np.sqrt(coef_Deter_simple)
print("Coeficiente de Determinación (R²):", coef_Deter_simple)
print("Coeficiente de Correlación (r):", coef_Correl_simple)
```

## Gráfica del modelo

```python
#Diagrama de dispersión de los datos reales
figure1 = px.scatter(data_frame=df, x=Variable_x, y=Variable_y,
                     title='Modelo Lineal Simple')
#Agregamos la recta de regresión con los valores predichos
figure1.add_trace(go.Scatter(x=df[Variable_x], y=y_pred, mode='lines',
                             name='Recta de regresión'))
figure1.show()
```

## Valores reales vs predichos

```python
#Creamos una tabla con los valores reales, predichos y el residual
Tabla_pred = pd.DataFrame({"Real": df[Variable_y],
                           "Predicho": y_pred})
Tabla_pred["Residual"] = Tabla_pred["Real"] - Tabla_pred["Predicho"]
Tabla_pred.round(2)
```
